```python
# Importaciones
from ddgs import DDGS
from thefuzz import fuzz
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
from lxml import html # <-- Asegúrate de tener esta importación


from selenium.common.exceptions import TimeoutException, NoSuchElementException, ElementClickInterceptedException

import time

def verify_and_scrape_siia(driver, name_to_verify, url):
    """
    Navega a una URL, cierra el MODAL HTML si aparece, verifica nombre y extrae datos.
    """
    try:
        # 1. NAVEGAR A LA URL
        driver.get(url)

        # --- GESTIÓN DEL MODAL (HTML / BOOTSTRAP) ---
        try:
            # Buscamos el botón "Aceptar" dentro del footer del modal.
            # Usamos WebDriverWait por si el modal tarda un segundo en aparecer (animación fade-in).
            boton_aceptar = WebDriverWait(driver, 3).until(
                EC.element_to_be_clickable((By.XPATH, "//div[@class='modal-footer']//button[contains(text(), 'Aceptar')]"))
            )
            print("    ⚠️ Modal detectado. Intentando cerrar...")
            boton_aceptar.click()
            
            # Esperamos un momento a que la animación de cierre (fade-out) termine
            # para que no bloquee el click en los elementos de abajo.
            time.sleep(1) 
            
        except TimeoutException:
            # Si en 3 segundos no aparece el botón, asumimos que no hay modal
            pass
        except ElementClickInterceptedException:
            # Si algo bloquea el click, intentamos forzarlo con Javascript
            driver.execute_script("arguments[0].click();", boton_aceptar)
        # ---------------------------------------------

        # 2. EXTRAER NOMBRE
        # Esperamos a que el H1 sea visible (ahora que el modal se fue)
        try:
            element_h1 = WebDriverWait(driver, 5).until(
                EC.visibility_of_element_located((By.XPATH, '/html/body/center/h1'))
            )
            scraped_name = element_h1.text
        except TimeoutException:
            print(f"    ❌ No se encontró el nombre (H1). ¿La página cargó bien?")
            return None

        # 3. VERIFICAR EL NOMBRE
        if fuzz.token_set_ratio(name_to_verify, scraped_name) < 90:
            print(f"    -> No coincide: '{name_to_verify}' vs '{scraped_name}'")
            return None 

        print(f"    ✅ Coincidencia encontrada para '{name_to_verify}'")

        # 4. EXTRAER DATOS
        data = {
            'name': scraped_name,
            'scopus': '',
            'orcid': '',
            'areas': []
        }

        # Scopus
        try:
            data['scopus'] = driver.find_element(By.XPATH, '/html/body/center/div/table[2]/tbody/tr[4]/td/span/a').get_attribute('href')
        except NoSuchElementException:
            pass

        # Orcid
        try:
            data['orcid'] = driver.find_element(By.XPATH, '/html/body/center/div/table[2]/tbody/tr[6]/td/span/a').get_attribute('href')
        except NoSuchElementException:
            pass

        # Áreas Temáticas
        try:
            td_element = driver.find_element(By.XPATH, "/html/body/center/div/table[3]/tbody/tr[2]/td")
            outer_html_td = td_element.get_attribute('outerHTML')
            tree = html.fromstring(outer_html_td)
            sub_elements = tree.xpath('./span')
            for sub_element in sub_elements:
                area_text = sub_element.text_content().strip()
                if area_text:
                    data['areas'].append(area_text)
        except Exception:
            pass 

        return data

    except Exception as e:
        print(f"    ❌ Error general en {url}: {e}")
        return None

# NUEVA FUNCIÓN COMBINADA
def verify_and_scrape_siia_old(driver, name_to_verify, url):
    """
    Navega a una URL, verifica si el nombre coincide y, si es así,
    extrae todos los datos requeridos. Devuelve los datos o None.
    """
    try:
        # 1. NAVEGAR A LA URL (UNA SOLA VEZ)
        driver.get(url)

        # 2. EXTRAER NOMBRE PARA VERIFICACIÓN
        scraped_name = driver.find_element(By.XPATH, '/html/body/center/h1').text

        # 3. VERIFICAR EL NOMBRE CON FUZZY MATCHING
        # Un ratio de > 90 suele ser más práctico que > 99
        if fuzz.token_set_ratio(name_to_verify, scraped_name) < 90:
            print(f"    -> No coincide: '{name_to_verify}' vs '{scraped_name}'")
            return None # Si el nombre no coincide, no seguimos

        print(f"    ✅ Coincidencia encontrada para '{name_to_verify}'")

        # 4. SI COINCIDE, EXTRAER EL RESTO DE LOS DATOS
        data = {
            'name': scraped_name,
            'scopus': '',
            'orcid': '',
            'areas': []
        }

        # Extraer Scopus
        try:
            data['scopus'] = driver.find_element(By.XPATH, '/html/body/center/div/table[2]/tbody/tr[4]/td/span/a').get_attribute('href')

        except NoSuchElementException:
            print("    ℹ️ No se encontró Scopus.")

        # Extraer Orcid
        try:
            data['orcid'] = driver.find_element(By.XPATH, '/html/body/center/div/table[2]/tbody/tr[6]/td/span/a').get_attribute('href')
        except NoSuchElementException:
            print("    ℹ️ No se encontró ORCID.")

        # Extraer Áreas Temáticas
        try:
            td_element = driver.find_element(By.XPATH, "/html/body/center/div/table[3]/tbody/tr[2]/td")
            outer_html_td = td_element.get_attribute('outerHTML')
            tree = html.fromstring(outer_html_td)
            sub_elements = tree.xpath('./span') # Es más específico buscar por <span>
            for sub_element in sub_elements:
                area_text = sub_element.text_content().strip()
                if area_text:
                    data['areas'].append(area_text)
        except Exception as e:
            print(f"    ❌ No se pudieron extraer las áreas. Error: {e}")

        return data

    except (TimeoutException, NoSuchElementException) as e:
        print(f"    ❌ Error al procesar la URL {url}: {e}")
        return None

def scrape_siia(driver, name_to_verify, url):
  """
  Navega a una URL, verifica si el nombre coincide y, si es así,
  extrae todos los datos requeridos. Devuelve los datos o None.
  """
  try:
      # 1. NAVEGAR A LA URL (UNA SOLA VEZ)
      driver.get(url)


      # 4. SI COINCIDE, EXTRAER EL RESTO DE LOS DATOS
      data = {
          'scopus': '',
          'orcid': '',
          'areas': []
      }

      # Extraer Scopus
      try:
          data['scopus'] = []
          td_element = driver.find_element(By.XPATH, "/html/body/center/div/table[2]/tbody/tr[4]/td")
          outer_html_td = td_element.get_attribute('outerHTML')
          tree = html.fromstring(outer_html_td)
          sub_elements = tree.xpath('./span') # Es más específico buscar por <span>
          for sub_element in sub_elements:
              data['scopus'].append(sub_element.text_content().strip())
      except NoSuchElementException:
          print("    ℹ️ No se encontró Scopus.")

      # Extraer Orcid
      try:
          data['orcid'] = driver.find_element(By.XPATH, '/html/body/center/div/table[2]/tbody/tr[6]/td/span/a').get_attribute('href')
      except NoSuchElementException:
          print("    ℹ️ No se encontró ORCID.")

      # Extraer Áreas Temáticas
      try:
          td_element = driver.find_element(By.XPATH, "/html/body/center/div/table[3]/tbody/tr[2]/td")
          outer_html_td = td_element.get_attribute('outerHTML')
          tree = html.fromstring(outer_html_td)
          sub_elements = tree.xpath('./span') # Es más específico buscar por <span>
          for sub_element in sub_elements:
              area_text = sub_element.text_content().strip()
              if area_text:
                  data['areas'].append(area_text)
      except Exception as e:
          print(f"    ❌ No se pudieron extraer las áreas. Error: {e}")

      return data

  except (TimeoutException, NoSuchElementException) as e:
      print(f"    ❌ Error al procesar la URL {url}: {e}")
      return None
```

```python
# --- Configuración del driver ---
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
driver = webdriver.Chrome(options=options)

profesores = {}

for profesor_name in lista:
  print(f"\nBuscando a: {profesor_name}")
  profesor_name = limpiar_nombre(profesor_name)
  profesores[profesor_name]={}
        
  siia_results = buscar_en_siia_serpapi(profesor_name)

  if not siia_results:
      print("    -> No se encontraron resultados en la búsqueda.")
      profesores[profesor_name]['siia'] = 'No encontrado'
      continue

  found = False
  for res in siia_results:
    scraped_data = verify_and_scrape_siia(driver, profesor_name, res['link'])
    if scraped_data:
      # Si la función devuelve datos, los guardamos
      profesores[profesor_name]['siia'] = res['link']
      profesores[profesor_name]['scopus'] = scraped_data['scopus']
      profesores[profesor_name]['orcid'] = scraped_data['orcid']
      profesores[profesor_name]['areas'] = scraped_data['areas']
      found = True
      break # Rompemos el bucle de resultados y pasamos al siguiente profesor

  if not found:
    profesores[profesor_name]['siia'] = 'Not found'



# Imprimir los resultados finales
import json
print(json.dumps(profesores, indent=2, ensure_ascii=False))
```
